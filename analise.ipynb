## Pergunta 1: quantos pedidos foram feitos em cada mês?


```python
pd.read_sql("""
SELECT strftime('%Y-%m', order_purchase_timestamp) AS mes,   -- tira o ano e o mês do texto da data
       COUNT(*) AS total_pedidos                             -- conta os pedidos de cada grupo
FROM orders
GROUP BY MES                                             -- um grupo para cada mês
ORDER BY MES DESC                                               -- do mês mais antigo para o mais novo
""", conexao)
```
